```python
!pip install -q "transformers>=4.51.0" "datasets>=2.19.0" "accelerate>=0.30.0" pillow pyyaml matplotlib

import os
import sys
import time
import json
import zipfile
import urllib.request
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor

REPO_URL = "https://github.com/syedmoazzamrazagilani/ATML-PA3.git"
REPO_DIR = Path("/kaggle/working/repo")

if not REPO_DIR.exists():
    !git clone {REPO_URL} {REPO_DIR}
else:
    !cd {REPO_DIR} && git fetch origin && git reset --hard origin/main

if str(REPO_DIR) not in sys.path:
    sys.path.insert(0, str(REPO_DIR))
%cd {REPO_DIR}

DATA_DIR = Path("/kaggle/working/data/task2_raw")
DATA_DIR.mkdir(parents=True, exist_ok=True)
IMG_DIR = Path("/kaggle/working/data/coco_images")
(IMG_DIR / "train2014").mkdir(parents=True, exist_ok=True)
(IMG_DIR / "val2014").mkdir(parents=True, exist_ok=True)

ARCHIVES = [
    (
        "coco_ann",
        "http://images.cocodataset.org/annotations/annotations_trainval2014.zip",
        "*captions_train2014*.json",
    ),
    (
        "vqa_train_q",
        "https://s3.amazonaws.com/cvmlp/vqa/mscoco/vqa/v2_Questions_Train_mscoco.zip",
        "*OpenEnded_mscoco_train2014_questions*.json",
    ),
    (
        "vqa_train_ann",
        "https://s3.amazonaws.com/cvmlp/vqa/mscoco/vqa/v2_Annotations_Train_mscoco.zip",
        "*mscoco_train2014_annotations*.json",
    ),
    (
        "vqa_train_pairs",
        "https://s3.amazonaws.com/cvmlp/vqa/mscoco/vqa/v2_Complementary_Pairs_Train_mscoco.zip",
        "*v2_mscoco_train2014_complementary_pairs*.json",
    ),
    (
        "vqa_val_q",
        "https://s3.amazonaws.com/cvmlp/vqa/mscoco/vqa/v2_Questions_Val_mscoco.zip",
        "*OpenEnded_mscoco_val2014_questions*.json",
    ),
    (
        "vqa_val_ann",
        "https://s3.amazonaws.com/cvmlp/vqa/mscoco/vqa/v2_Annotations_Val_mscoco.zip",
        "*mscoco_val2014_annotations*.json",
    ),
    (
        "vqa_val_pairs",
        "https://s3.amazonaws.com/cvmlp/vqa/mscoco/vqa/v2_Complementary_Pairs_Val_mscoco.zip",
        "*v2_mscoco_val2014_complementary_pairs*.json",
    ),
]

print("--- Step 1: Downloading Annotations and Question Pairs ---")
for name, url, expected_glob in ARCHIVES:
    if not list(DATA_DIR.rglob(expected_glob)):
        zip_path = DATA_DIR / f"{name}.zip"
        print(f"Downloading {name} from {url} ...")
        urllib.request.urlretrieve(url, zip_path)
        with zipfile.ZipFile(zip_path, "r") as zf:
            zf.extractall(DATA_DIR)
        zip_path.unlink()
    else:
        print(f"Already extracted: {name}")

print("All 7 annotation archives verified on disk!")

from task2.data_prep import build_task2_splits

task2_out_dir = Path("/kaggle/working/task2_outputs")
task2_out_dir.mkdir(parents=True, exist_ok=True)

splits = build_task2_splits(
    coco_train_captions_json=list(DATA_DIR.rglob("*captions_train2014*.json"))[0],
    coco_val_captions_json=list(DATA_DIR.rglob("*captions_val2014*.json"))[0],
    coco_train_instances_json=list(DATA_DIR.rglob("*instances_train2014*.json"))[0],
    coco_val_instances_json=list(DATA_DIR.rglob("*instances_val2014*.json"))[0],
    vqa_train_questions_json=list(DATA_DIR.rglob("*OpenEnded_mscoco_train2014_questions*.json"))[0],
    vqa_train_annotations_json=list(DATA_DIR.rglob("*mscoco_train2014_annotations*.json"))[0],
    vqa_train_comp_pairs_json=list(DATA_DIR.rglob("*v2_mscoco_train2014_complementary_pairs*.json"))[0],
    vqa_val_questions_json=list(DATA_DIR.rglob("*OpenEnded_mscoco_val2014_questions*.json"))[0],
    vqa_val_annotations_json=list(DATA_DIR.rglob("*mscoco_val2014_annotations*.json"))[0],
    vqa_val_comp_pairs_json=list(DATA_DIR.rglob("*v2_mscoco_val2014_complementary_pairs*.json"))[0],
    output_dir=task2_out_dir,
    seed=6304,
)

print("--- Step 2: Downloading Selected Images ---")
def download_single_image(args):
    split, iid = args
    fname = f"COCO_{split}_{int(iid):012d}.jpg"
    dest = IMG_DIR / split / fname
    if dest.exists() and dest.stat().st_size > 0:
        return True
    url = f"http://images.cocodataset.org/{split}/{fname}"
    for attempt in range(3):
        try:
            urllib.request.urlretrieve(url, dest)
            if dest.exists() and dest.stat().st_size > 0:
                return True
        except Exception:
            time.sleep(0.5 * (attempt + 1))
    return False

tasks = [("train2014", iid) for iid in splits["image_ids"]["train2014"]] + \
        [("val2014", iid) for iid in (splits["image_ids"]["val2014_dev"] + splits["image_ids"]["val2014_test"])]

print(f"Fetching {len(tasks)} target images...")
with ThreadPoolExecutor(max_workers=16) as pool:
    results = list(pool.map(download_single_image, tasks))

print(f"Downloaded {sum(results)}/{len(tasks)} images successfully.")
assert all(results), f"Failed to download {len(tasks) - sum(results)} images!"
!du -sh /kaggle/working/data/coco_images/*
```

```python
import os
import shutil
from pathlib import Path
from IPython.display import FileLink, display

out_dir = Path("/kaggle/working/task2_outputs")
bundle_dir = Path("/kaggle/working/task2_report_bundle")
shutil.rmtree(bundle_dir, ignore_errors=True)
bundle_dir.mkdir(parents=True, exist_ok=True)

for fname in ["task2_splits_and_ids.json", "siglip_features_fp16.stats.json"]:
    if (out_dir / fname).exists():
        shutil.copy2(out_dir / fname, bundle_dir / fname)

for run_name in ["run_196tokens", "run_49tokens"]:
    r_src = out_dir / run_name
    r_dst = bundle_dir / run_name
    if r_src.exists():
        r_dst.mkdir(parents=True, exist_ok=True)
        for log_file in ["train_log.jsonl", "dev_eval_log.jsonl", "training_summary.json"]:
            if (r_src / log_file).exists():
                shutil.copy2(r_src / log_file, r_dst / log_file)
        for sub in ["eval", "figures"]:
            if (r_src / sub).exists():
                shutil.copytree(r_src / sub, r_dst / sub, dirs_exist_ok=True)

os.chdir("/kaggle/working")
zip_base = "/kaggle/working/task2_report_artifacts"
shutil.make_archive(zip_base, "zip", bundle_dir)
shutil.rmtree(bundle_dir, ignore_errors=True)

zip_file = Path("task2_report_artifacts.zip")
size_mb = zip_file.stat().st_size / (1024 ** 2)
print("\n" + "=" * 60)
print(f"Task 2 artifacts packaged: {zip_file.name} ({size_mb:.2f} MB)")
print("=" * 60 + "\n")
display(FileLink("task2_report_artifacts.zip"))
```
