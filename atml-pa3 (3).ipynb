```python
!pip install -q "transformers>=4.51.0" "datasets>=2.19.0" "accelerate>=0.30.0" "math-verify" gdown pyarrow pyyaml matplotlib pandas tqdm

import os
import sys
import urllib.request
from pathlib import Path
import gdown

REPO_URL = "https://github.com/syedmoazzamrazagilani/ATML-PA3.git"
REPO_DIR = Path("/kaggle/working/repo")

if not REPO_DIR.exists():
    !git clone {REPO_URL} {REPO_DIR}
else:
    !cd {REPO_DIR} && git fetch origin && git reset --hard origin/main

if str(REPO_DIR) not in sys.path:
    sys.path.insert(0, str(REPO_DIR))
%cd {REPO_DIR}

DATA_ROOT = Path("/kaggle/working/data/task3")
DATA_ROOT.mkdir(parents=True, exist_ok=True)

if not list(DATA_ROOT.rglob("generations.parquet")):
    print("Downloading ATML-PA3-Data from Google Drive...")
    gdown.download_folder(
        id="1utX-Gfi9vvaoi1mi0pzNpDGL1o1fAh23",
        output=str(DATA_ROOT),
        quiet=False,
        use_cookies=False,
    )

meta_path = DATA_ROOT / "testmini.json"
if not meta_path.exists():
    print("Downloading MathVista testmini.json...")
    urllib.request.urlretrieve(
        "https://raw.githubusercontent.com/lupantech/MathVista/main/data/testmini.json",
        meta_path,
    )

print("Verified parquet files:", [str(p.relative_to(DATA_ROOT)) for p in DATA_ROOT.rglob("*.parquet")])
```

```python
import os
import shutil
from pathlib import Path
from IPython.display import FileLink, display

out_dir = Path("/kaggle/working/task3_outputs")
bundle_dir = Path("/kaggle/working/task3_report_bundle")
shutil.rmtree(bundle_dir, ignore_errors=True)
bundle_dir.mkdir(parents=True, exist_ok=True)

for sub in ["step1", "step2", "step3", "step4", "step5", "step6", "figures", "tables"]:
    if (out_dir / sub).exists():
        shutil.copytree(out_dir / sub, bundle_dir / sub, dirs_exist_ok=True)

if (out_dir / "task3_full_manifest.json").exists():
    shutil.copy2(out_dir / "task3_full_manifest.json", bundle_dir / "task3_full_manifest.json")

os.chdir("/kaggle/working")
zip_base = "/kaggle/working/task3_report_artifacts"
shutil.make_archive(zip_base, "zip", bundle_dir)
shutil.rmtree(bundle_dir, ignore_errors=True)

zip_file = Path("task3_report_artifacts.zip")
size_mb = zip_file.stat().st_size / (1024 ** 2)
print("\n" + "=" * 60)
print(f"Task 3 artifacts packaged: {zip_file.name} ({size_mb:.2f} MB)")
print("=" * 60 + "\n")
display(FileLink("task3_report_artifacts.zip"))
```
